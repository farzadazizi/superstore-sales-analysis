```sql

WITH yearly_sales AS (
    SELECT
        YEAR(`Order Date`) AS sales_year,
        SUM(Sales) AS total_sales,
        SUM(Profit) AS total_profit
    FROM spotify_catalog.superstore.sample_superstore
    GROUP BY YEAR(`Order Date`)
),

sales_growth AS (
    SELECT
        *,
        LAG(total_sales) OVER (
            ORDER BY sales_year
        ) AS previous_year_sales,

        LAG(total_profit) OVER (
            ORDER BY sales_year
        ) AS previous_year_profit
    FROM yearly_sales
)

SELECT
    sales_year,
    ROUND(total_sales, 2) AS total_sales,

    ROUND(
        (total_sales - previous_year_sales)
        / previous_year_sales * 100, 2
    ) AS sales_growth_pct,

    ROUND(total_profit, 2) AS total_profit,

    ROUND(
        (total_profit - previous_year_profit)
        / previous_year_profit * 100, 2
    ) AS profit_growth_pct,

    ROUND(
        total_profit / total_sales * 100, 2
    ) AS profit_margin_pct

FROM sales_growth
ORDER BY sales_year;
```

```sql

WITH yearly_sales AS (
    SELECT
        YEAR(`Order Date`) AS sales_year,
        SUM(Sales) AS total_sales,
        SUM(Profit) AS total_profit
    FROM spotify_catalog.superstore.sample_superstore
    GROUP BY YEAR(`Order Date`)
),

sales_growth AS (
    SELECT
        *,
        LAG(total_sales) OVER (
            ORDER BY sales_year
        ) AS previous_year_sales,

        LAG(total_profit) OVER (
            ORDER BY sales_year
        ) AS previous_year_profit
    FROM yearly_sales
)

SELECT
    sales_year,
    ROUND(total_sales, 2) AS total_sales,

    ROUND(
        (total_sales - previous_year_sales)
        / previous_year_sales * 100, 2
    ) AS sales_growth_pct,

    ROUND(total_profit, 2) AS total_profit,

    ROUND(
        (total_profit - previous_year_profit)
        / previous_year_profit * 100, 2
    ) AS profit_growth_pct,

    ROUND(
        total_profit / total_sales * 100, 2
    ) AS profit_margin_pct

FROM sales_growth
ORDER BY sales_year;
```

```sql

SELECT
    `Sub-Category`,

    ROUND(SUM(Sales), 2) AS total_sales,

    ROUND(SUM(Profit), 2) AS total_profit,

    ROUND(
        SUM(Profit) / SUM(Sales) * 100, 2
    ) AS profit_margin_pct,

    ROUND(AVG(Discount) * 100, 2) AS avg_discount_pct,

    SUM(Quantity) AS total_quantity

FROM spotify_catalog.superstore.sample_superstore

WHERE Category = 'Furniture'

GROUP BY `Sub-Category`

ORDER BY profit_margin_pct ASC;
```

```sql

SELECT
    `Sub-Category`,

    CASE
        WHEN Discount = 0 THEN '0%'
        WHEN Discount <= 0.10 THEN '1-10%'
        WHEN Discount <= 0.20 THEN '11-20%'
        WHEN Discount <= 0.30 THEN '21-30%'
        ELSE 'Above 30%'
    END AS discount_range,

    COUNT(*) AS total_records,

    ROUND(SUM(Sales), 2) AS total_sales,

    ROUND(SUM(Profit), 2) AS total_profit,

    ROUND(
        SUM(Profit) / SUM(Sales) * 100, 2
    ) AS profit_margin_pct

FROM spotify_catalog.superstore.sample_superstore

WHERE `Sub-Category` IN ('Tables', 'Bookcases')

GROUP BY
    `Sub-Category`,
    CASE
        WHEN Discount = 0 THEN '0%'
        WHEN Discount <= 0.10 THEN '1-10%'
        WHEN Discount <= 0.20 THEN '11-20%'
        WHEN Discount <= 0.30 THEN '21-30%'
        ELSE 'Above 30%'
    END

ORDER BY
    `Sub-Category`,
    discount_range;
```

```sql
SELECT
    `Sub-Category`,
    CASE
        WHEN Discount = 0 THEN '0%'
        WHEN Discount <= 0.10 THEN '1-10%'
        WHEN Discount <= 0.20 THEN '11-20%'
        WHEN Discount <= 0.30 THEN '21-30%'
        ELSE 'Above 30%'
    END AS discount_range,

    COUNT(*) AS total_records,
    ROUND(SUM(Sales), 2) AS total_sales,
    ROUND(SUM(Profit), 2) AS total_profit,

    ROUND(
        SUM(Profit) / NULLIF(SUM(Sales), 0) * 100,
        2
    ) AS profit_margin_pct

FROM spotify_catalog.superstore.sample_superstore

WHERE Region = 'Central'
  AND Category = 'Furniture'

GROUP BY
    `Sub-Category`,
    CASE
        WHEN Discount = 0 THEN '0%'
        WHEN Discount <= 0.10 THEN '1-10%'
        WHEN Discount <= 0.20 THEN '11-20%'
        WHEN Discount <= 0.30 THEN '21-30%'
        ELSE 'Above 30%'
    END

ORDER BY
    `Sub-Category`,
    discount_range;
```
